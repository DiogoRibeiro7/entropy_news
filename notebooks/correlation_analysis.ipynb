# Correlation Analysis Notebook
This notebook demonstrates how to compute rolling correlations between entropy scores and market returns using the helper utilities.

```python
from entropy_news.utils import correlation, rolling_correlation
import pandas as pd

news = pd.Series([0.12, 0.08, 0.05, 0.09, 0.03])
returns = pd.Series([0.10, 0.06, 0.04, 0.08, 0.02])
overall = correlation(news, returns)
rolling = rolling_correlation(news, returns, window=3)
print('Overall correlation:', overall)
print('Rolling correlation:', rolling.tolist())
```
